# TM-60 and CM-60 tSNEs

```python
data = pd.read_csv(r'M:\ML_scripts\chemplot\tobramycin_full_labelled.csv').drop(columns=['Unnamed: 0'])          # full meaning both training and testing data (for tob -> 524 + 1178)
data_col = pd.read_csv(r'M:\ML_scripts\chemplot\colistin_full_labelled.csv').drop(columns=['Unnamed: 0'])        # for col -> 890 + 1146

tob_train = data[data['set'] == 'train']
tob_test = data[data['set'] == 'test']

col_train = data_col[data_col['set'] == 'train']
col_test = data_col[data_col['set'] == 'test']
```

```python
sns.set_style('white')

chemplot_data = Plotter.from_smiles(data['Structure'], target=data['set'], target_type='C', sim_type='structural');
chemplot_data.tsne(perplexity=20, random_state=0)
chemplot_data.visualize_plot(size=10, kind='scatter', title='TM-60 tSNE');

col_chemplot_data = Plotter.from_smiles(data_col['Structure'], target=data_col['set'], target_type='C', sim_type='structural');
col_chemplot_data.tsne(perplexity=20, random_state=0)
col_chemplot_data.visualize_plot(size=10, kind='scatter', title='CM-60 tSNE');
```
