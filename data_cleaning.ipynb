# Data cleaning

```python
import pandas as pd 
import numpy as np

data = pd.read_csv(r"M:\ML_scripts\data\N1426-1.csv")  # load the original data
structure = data['Structure']   # extract structure and activities
tob_activity = data['Tob_Pot_1']
col_activity = data['Col_Pot_1']

df = pd.DataFrame()
df.insert(0, 'Structure', structure)
df.insert(1, 'Tob_Pot_1', tob_activity) # create dataframe with structure and activities
df.insert(2, 'Col_Pot_1', col_activity)

tob_df = df.drop(columns=['Col_Pot_1']) # drop colistin activity
col_df = df.drop(columns=['Tob_Pot_1']) # drop tobramycin activity

tob_df.insert(1, 'Class', 0)    #    insert blank columns for binary classification
col_df.insert(1, 'Class', 0)

for i in range(len(tob_df)):
    if tob_df.loc[i, 'Tob_Pot_1'] > 60:     # if the activity is greater than 60, classify as active (1) 
        tob_df.loc[i, 'Class'] = 1          # otherwise classify as inactive (0)
    else:
        tob_df.loc[i, 'Class'] = 0

for j in range(len(col_df)):
    if col_df.loc[j, 'Col_Pot_1'] > 60:     # compounds with 60.0 are considered 0 
        col_df.loc[j, 'Class'] = 1          # only 2 compounds with 60.0 in the data (1 in each)      
    else:
        col_df.loc[j, 'Class'] = 0

tob_df.convert_dtypes()
col_df.convert_dtypes()

tob_df = tob_df.dropna()        # drop empty rows
col_df = col_df.dropna()

tob_df = tob_df.reset_index(drop=True)
col_df = col_df.reset_index(drop=True)

tob_active_count = 0        # section to count the active and inactive compounds
col_active_count = 0

for x in tob_df['Class']:
    if x == 1:
        tob_active_count += 1
for y in col_df['Class']:
    if y == 1:
        col_active_count += 1

print('Active compounds for tobramycin:', tob_active_count)
print('Inactive compounds for tobramycin:', len(tob_df)-tob_active_count)

print('\nActive compounds for colistin:', col_active_count)
print('Inactive compounds for colistin:', len(col_df)-col_active_count)
```
